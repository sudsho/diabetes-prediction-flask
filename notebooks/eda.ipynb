## Zero-as-missing

Glucose, BloodPressure, SkinThickness, Insulin and BMI cannot really be 0.
Count how often that shows up.

```python
cols = ['Glucose','BloodPressure','SkinThickness','Insulin','BMI']
(df[cols] == 0).sum()
```

```python
df[cols].replace(0, np.nan).hist(bins=30, figsize=(10,6))
plt.tight_layout()
```
